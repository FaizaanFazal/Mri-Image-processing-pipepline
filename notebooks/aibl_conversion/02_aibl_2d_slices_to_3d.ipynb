# aibl 2d slices to 3d images

this notebook combines each aibl dicom series into one compressed raw 3d nifti image under `Data/AIBL/raw3DAll/T1`, `T2_FLAIR`, or `SWI`. it keeps sequence description and modality family in a resumable manifest under `Data/AIBL/manifests` with the original 2d folder and converted 3d path. notebook output is aggregate only and does not show ids or raw paths.

```python
from pathlib import Path
import sys
from IPython.display import display
```

```python
from pathlib import Path
import sys
from IPython.display import display

candidates = [Path.cwd(), *Path.cwd().parents, Path.cwd() / 'SkullStrippingProject']
PROJECT_ROOT = next(path for path in candidates if (path / 'src' / 'skullstrip_pipeline').exists())
sys.path.insert(0, str(PROJECT_ROOT / 'src'))
from skullstrip_pipeline.aibl_conversion import run_aibl_conversion, save_conversion_qc_examples

MAX_FILES = None #none for all 
REPLACE = False
CPU_WORKERS = 1
STOP_ON_ERROR = False
print(f'test settings: max files={MAX_FILES}, replace={REPLACE}, cpu workers={CPU_WORKERS}')
```

```python
conversion_manifest = run_aibl_conversion(max_files=MAX_FILES, replace=REPLACE, workers=CPU_WORKERS, stop_on_error=STOP_ON_ERROR)
status_counts = conversion_manifest.groupby(['status', 'qc_pass'], dropna=False).size().rename('count').reset_index()
display(status_counts)
```
